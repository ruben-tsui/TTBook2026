### 下載 `Open Subtitles 2024` / Download `Open Subtitles 2024`

```python
import polars as pl
from IPython.display import Markdown, display, HTML# Read an Apache Parquet file
!gdown 1kyoglv4n2qwzjt7REULHegw_35mg8SsZ -O OpenSubtitles2024.parquet

Corpus = 'OpenSubtitles2024'
df = pl.read_parquet(f"/content/{Corpus}.parquet")
```

### 開始搜尋（以中文或英文）/ Start Searching Corpus with English or Chinese phrase/regex

```python
%%time

Corpus_語料庫 = 'OpenSubtitles2024' # @param ['OpenSubtitles2024']
corpus = Corpus_語料庫

Language_語言 = "en" # @param ["en", "zh"]
language = Language_語言

Case_sensitive_區分大小寫 = False # @param {type:"boolean"}
case_sensitive = Case_sensitive_區分大小寫

Search_en_以英文搜尋 = " chip on \\S+ shoulder" # @param {"type":"string"}
search_en = Search_en_以英文搜尋

Search_zh_以中文搜尋 = "發酒瘋" # @param {"type":"string"}
search_zh = Search_zh_以中文搜尋

# search on this column
column = language

Max_Results = 50 # @param

if column == "en":
    if case_sensitive:
        search_str = search_en
    else: # case insensitive
        search_str = '(?i)' + search_en
    search_str = search_en
elif column == "zh":
    search_str = search_zh
res = df.filter(
    pl.col(column).str.contains(search_str, literal=False) # use regex by default
)
query_results = res.to_dict(as_series=False)
length =  len(query_results[column])
print("No. of Matches:", length)
MAX_ROWS = min(length, Max_Results)
rows = []
for i in range(length):
    #print(i)
    e = query_results['en'][i]
    en_sentence = f"<span style='font-size: 18px; font-family: Consolas'>{e}</span>"
    z = query_results['zh'][i]
    zh_sentence = f"<span style='font-size: 18px; font-family: 'Microsoft Jhenghei'>{z}</span>"

    rows.append(f"""
    <tr>
    <td width='20px'>{i}</td>
    <td width='500px'>{en_sentence}</td>
    <td width='300px'>{zh_sentence}</td>
    </tr>
    """)
    if i >= MAX_ROWS: break  # Maximum no. of rows returned

whole_table = f"""
<table border="1" cellspacing="0" cellpadding="4">
{'\n'.join(rows)}
</table>
"""
display(HTML(whole_table))
```
